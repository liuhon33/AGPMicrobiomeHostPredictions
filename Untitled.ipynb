```python
# ============================================================
# 6a ONLY: Re-create ordered PC outer-fold CV R² plot
# without rerunning PCA/RF training
# ============================================================

import os
import json
import joblib
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from IPython.display import display, FileLink

# ------------------------------------------------------------
# 1) Paths
# ------------------------------------------------------------
output_dir = Path("./hongrui_result/RF_on_PCA_Results/")
output_dir.mkdir(parents=True, exist_ok=True)

model_dir = Path("./RF_PCA_Trained_Models/")

# ------------------------------------------------------------
# 2) Try to load nested_cv_results if missing OR currently None
# ------------------------------------------------------------

candidate_paths = [
    output_dir / "nested_cv_results.joblib",
    output_dir / "nested_cv_results.pkl",
    output_dir / "nested_cv_results.json",
    model_dir / "nested_cv_results.joblib",
    model_dir / "nested_cv_results.pkl",
    Path("nested_cv_results.joblib"),
    Path("nested_cv_results.pkl"),
    Path("nested_cv_results.json"),
]

# Important fix:
# The previous version only checked whether the variable existed.
# Here we also check whether it is None.
if globals().get("nested_cv_results", None) is None:
    print("nested_cv_results is missing or None. Searching for a saved copy...")

    nested_cv_results = None
    found_path = None

    for path in candidate_paths:
        if path.exists():
            print(f"Found candidate file: {path}")
            found_path = path

            if path.suffix == ".joblib":
                nested_cv_results = joblib.load(path)

            elif path.suffix in [".pkl", ".pickle"]:
                with open(path, "rb") as f:
                    nested_cv_results = pickle.load(f)

            elif path.suffix == ".json":
                with open(path, "r") as f:
                    nested_cv_results = json.load(f)

            break

    if nested_cv_results is None:
        searched = "\n".join([f"  - {p}" for p in candidate_paths])

        raise FileNotFoundError(
            "nested_cv_results is still None.\n\n"
            "This means the exact fold-level results for part 6.a were not found.\n"
            "The final saved RF models are not enough to recreate the exact outer-fold CV plot.\n\n"
            "Searched these locations:\n"
            f"{searched}\n\n"
            "If you still have the notebook session where nested_cv_results was created, run:\n\n"
            "joblib.dump(nested_cv_results, './hongrui_result/RF_on_PCA_Results/nested_cv_results.joblib')\n\n"
            "Then rerun this plotting block."
        )

else:
    print("Using nested_cv_results already in notebook memory.")

# ------------------------------------------------------------
# 3) Validate that fold-level R² values exist
# ------------------------------------------------------------

if not isinstance(nested_cv_results, dict):
    raise TypeError(
        f"nested_cv_results should be a dictionary, but got {type(nested_cv_results)}."
    )

if len(nested_cv_results) == 0:
    raise ValueError("nested_cv_results is an empty dictionary.")

example_pc = next(iter(nested_cv_results))

required_keys = {"avg_r2", "fold_r2"}
missing_keys = required_keys - set(nested_cv_results[example_pc].keys())

if missing_keys:
    raise KeyError(
        f"nested_cv_results exists, but it is missing: {missing_keys}\n"
        "The exact 6.a plot needs fold-level R² values stored as fold_r2."
    )

outer_cv_splits = globals().get(
    "outer_cv_splits",
    len(nested_cv_results[example_pc]["fold_r2"])
)
```

```python
# ------------------------------------------------------------
# 4) Recreate 6.a plot
# ------------------------------------------------------------
print("\n--- Recreating 6a. Ordered PC outer-fold CV R² plot ---")

r2_threshold = 0.02
rng = np.random.RandomState(0)

# Sort PCs by mean R², low to high
pc_order = sorted(
    nested_cv_results.keys(),
    key=lambda k: nested_cv_results[k]["avg_r2"]
)

means = np.array(
    [nested_cv_results[pc]["avg_r2"] for pc in pc_order],
    dtype=float
)

n_total = len(pc_order)
selected_mask = means > r2_threshold
n_selected = int(selected_mask.sum())

idxs = np.where(selected_mask)[0]
cross_idx = int(idxs[0]) if len(idxs) > 0 else None

fig, ax = plt.subplots(figsize=(16, 7.6), dpi=300)

x = np.arange(n_total)

# Fold-level points
for i, pc in enumerate(pc_order):
    fold_scores = np.array(nested_cv_results[pc]["fold_r2"], dtype=float)
    jitter = rng.uniform(-0.18, 0.18, size=len(fold_scores))

    if nested_cv_results[pc]["avg_r2"] > r2_threshold:
        point_color = "gold"
        point_alpha = 0.80
        point_size = 16
    else:
        point_color = "steelblue"
        point_alpha = 0.45
        point_size = 13

    ax.scatter(
        i + jitter,
        fold_scores,
        alpha=point_alpha,
        s=point_size,
        color=point_color,
        edgecolor="none",
        rasterized=True
    )

# Mean R² line
ax.plot(
    x,
    means,
    marker="o",
    markersize=3.5,
    linewidth=2.2,
    color="black",
    label="PC mean outer-CV $R^2$",
    zorder=5
)

# Threshold line
ax.axhline(
    r2_threshold,
    linestyle="--",
    linewidth=1.6,
    color="black",
    label=f"Threshold = {r2_threshold:.2f}"
)

# Vertical cutoff guide
if cross_idx is not None:
    ax.axvline(
        cross_idx - 0.5,
        linestyle="--",
        linewidth=1.1,
        color="black"
    )

ax.set_title(
    "Ordered PC outer-fold CV $R^2$ scores",
    fontsize=16,
    pad=8
)

ax.set_xlabel(
    "PCs ordered by outer-CV mean $R^2$ (low \u2192 high)",
    fontsize=14,
    labelpad=6
)

ax.set_ylabel(
    "Outer-fold and mean $R^2$",
    fontsize=14,
    labelpad=6
)

ax.tick_params(axis="both", labelsize=12)

tick_step = 25 if n_total >= 100 else 10
ticks = np.arange(0, n_total, tick_step)
ax.set_xticks(ticks)
ax.set_xticklabels(ticks, fontsize=12)

ax.grid(True, linestyle="--", alpha=0.35)

y_all = []
for pc in pc_order:
    y_all.extend(nested_cv_results[pc]["fold_r2"])

y_all = np.array(y_all, dtype=float)

y_min = min(np.nanmin(y_all), np.nanmin(means))
y_max = max(np.nanmax(y_all), np.nanmax(means))
y_pad = 0.08 * (y_max - y_min)

ax.set_ylim(y_min - y_pad, y_max + y_pad)

info = (
    f"Threshold = {r2_threshold:.3f}\n"
    f"Selected PCs = {n_selected}/{n_total}\n"
    f"Outer folds per PC = {outer_cv_splits}"
)

ax.text(
    0.01,
    0.96,
    info,
    transform=ax.transAxes,
    va="top",
    ha="left",
    fontsize=10.5,
    bbox=dict(
        boxstyle="round,pad=0.3",
        fc="white",
        ec="gray",
        alpha=0.9
    )
)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

# Save both PDF and PNG
pdf_path = output_dir / "ordered_pc_outer_fold_cv_r2.pdf"
png_path = output_dir / "ordered_pc_outer_fold_cv_r2.png"

plt.savefig(pdf_path, bbox_inches="tight")
plt.savefig(png_path, bbox_inches="tight", dpi=300)

plt.show()
plt.close()

print(f"Saved PDF: {pdf_path}")
print(f"Saved PNG: {png_path}")

display(FileLink(str(pdf_path)))
display(FileLink(str(png_path)))

# ------------------------------------------------------------
# 5) Print top 15 PCs by mean outer-CV R²
# ------------------------------------------------------------
top15_df = (
    pd.DataFrame([
        {
            "pc": pc,
            "mean_r2": float(res.get("avg_r2", np.nan)),
            "sd_r2": float(res.get("std_r2", np.nan)),
        }
        for pc, res in nested_cv_results.items()
    ])
    .sort_values("mean_r2", ascending=False)
    .head(15)
    .reset_index(drop=True)
)

print("\n=== Top 15 PCs by mean outer-CV R² ===")
print(
    top15_df.to_string(
        index=False,
        justify="left",
        formatters={
            "mean_r2": "{:.4f}".format,
            "sd_r2": "{:.4f}".format
        }
    )
)
```

```python
# ============================================================
# PC1–PC10 Nested-CV R² Bar Plot
# - Isolated plotting block
# - Does NOT rerun nested CV
# - Uses nested_cv_results in memory OR loads saved joblib
# - Keeps natural order: PC1, PC2, ..., PC10
# ============================================================

import os
from pathlib import Path
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("\n--- Generating PC1–PC10 nested-CV R² plot ---")

# ------------------------------------------------------------
# 1) Define output directory safely
# ------------------------------------------------------------
if "output_dir" not in globals():
    output_dir = "./hongrui_result/RF_on_PCA_Results/"

output_dir = Path(output_dir)
output_dir.mkdir(parents=True, exist_ok=True)

# ------------------------------------------------------------
# 2) Load nested_cv_results if not already available
# ------------------------------------------------------------
nested_cv_results_in_memory = globals().get("nested_cv_results", None)

if not isinstance(nested_cv_results_in_memory, dict) or len(nested_cv_results_in_memory) == 0:
    cv_path = output_dir / "nested_cv_results.joblib"

    if not cv_path.exists():
        raise FileNotFoundError(
            f"nested_cv_results was not found in notebook memory, and saved file was not found at:\n"
            f"{cv_path}\n\n"
            "Please make sure the nested CV cell has been run once and that "
            "nested_cv_results.joblib was saved."
        )

    nested_cv_results = joblib.load(cv_path)
    print(f"Loaded nested_cv_results from: {cv_path}")

else:
    nested_cv_results = nested_cv_results_in_memory
    print("Using nested_cv_results already in notebook memory.")

# ------------------------------------------------------------
# 3) Extract PC1–PC10 only
# ------------------------------------------------------------
pcs_to_plot = [f"PC{i}" for i in range(1, 11)]

missing_pcs = [pc for pc in pcs_to_plot if pc not in nested_cv_results]

if missing_pcs:
    raise KeyError(
        f"The following PCs are missing from nested_cv_results: {missing_pcs}\n"
        "Check whether your nested CV block predicted at least PC1–PC10."
    )

records = []

for pc in pcs_to_plot:
    res = nested_cv_results[pc]

    # Prefer fold-level R² values if available
    if "fold_r2" in res:
        fold_scores = np.array(res["fold_r2"], dtype=float)
        mean_r2 = float(np.nanmean(fold_scores))
        sd_r2 = float(np.nanstd(fold_scores, ddof=1))
        n_folds = int(np.sum(~np.isnan(fold_scores)))

    # Fallback to summary values
    elif "avg_r2" in res and "std_r2" in res:
        mean_r2 = float(res["avg_r2"])
        sd_r2 = float(res["std_r2"])
        n_folds = np.nan

    else:
        raise KeyError(
            f"{pc} does not contain either 'fold_r2' or both 'avg_r2' and 'std_r2'.\n"
            f"Available keys are: {list(res.keys())}"
        )

    records.append({
        "PC": pc,
        "pc_number": int(pc.replace("PC", "")),
        "mean_r2": mean_r2,
        "sd_r2": sd_r2,
        "n_folds": n_folds
    })

pc10_r2_df = pd.DataFrame(records)

# ------------------------------------------------------------
# 4) Keep natural PC order: PC1, PC2, ..., PC10
# ------------------------------------------------------------
pc10_r2_df = (
    pc10_r2_df
    .sort_values("pc_number", ascending=True)
    .reset_index(drop=True)
)

print("\n=== PC1–PC10 outer-CV R² in natural PC order ===")
print(
    pc10_r2_df[["PC", "mean_r2", "sd_r2", "n_folds"]].to_string(
        index=False,
        formatters={
            "mean_r2": "{:.4f}".format,
            "sd_r2": "{:.4f}".format
        }
    )
)

# Save table
csv_path = output_dir / "pc1_pc10_ordered_nested_cv_r2.csv"
pc10_r2_df.to_csv(csv_path, index=False)
print(f"\nSaved PC1–PC10 R² table to: {csv_path}")

# ------------------------------------------------------------
# 5) Plot PC1–PC10 in natural order
# ------------------------------------------------------------
fig, ax = plt.subplots(figsize=(9, 5.5), dpi=300)

x = np.arange(len(pc10_r2_df))

bars = ax.bar(
    x,
    pc10_r2_df["mean_r2"],
    yerr=pc10_r2_df["sd_r2"],
    capsize=5,
    color="steelblue",
    edgecolor="black",
    alpha=0.85
)

# Horizontal reference line at zero
ax.axhline(
    0,
    color="black",
    linewidth=1
)

# Optional threshold line
# Change this to 0.03 if you want to match the stricter threshold.
r2_threshold = 0.02

ax.axhline(
    r2_threshold,
    linestyle="--",
    linewidth=1.2,
    color="gray",
    label=f"Threshold = {r2_threshold:.2f}"
)

# X-axis labels
ax.set_xticks(x)
ax.set_xticklabels(pc10_r2_df["PC"], fontsize=11)

# Axis labels and title
ax.set_ylabel("Mean outer-CV $R^2$ ± SD", fontsize=12)
ax.set_xlabel("Microbiome principal component", fontsize=12)
ax.set_title("Nested cross-validation prediction accuracy for PC1–PC10", fontsize=14)

# Styling
ax.grid(axis="y", linestyle="--", alpha=0.4)
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.legend(frameon=False)

# ------------------------------------------------------------
# 6) Add mean R² labels above bars
# ------------------------------------------------------------
y_min = min(0, (pc10_r2_df["mean_r2"] - pc10_r2_df["sd_r2"]).min())
y_max = max(0, (pc10_r2_df["mean_r2"] + pc10_r2_df["sd_r2"]).max())

y_pad = 0.08 * (y_max - y_min) if y_max > y_min else 0.01

for bar, mean_r2 in zip(bars, pc10_r2_df["mean_r2"]):
    if mean_r2 >= 0:
        y_text = bar.get_height() + y_pad * 0.25
        va = "bottom"
    else:
        y_text = bar.get_height() - y_pad * 0.25
        va = "top"

    ax.text(
        bar.get_x() + bar.get_width() / 2,
        y_text,
        f"{mean_r2:.3f}",
        ha="center",
        va=va,
        fontsize=9,
        fontweight="bold"
    )

ax.set_ylim(y_min - y_pad, y_max + y_pad * 1.8)

plt.tight_layout()

# ------------------------------------------------------------
# 7) Save figure
# ------------------------------------------------------------
pdf_path = output_dir / "pc1_pc10_ordered_nested_cv_r2_barplot.pdf"
png_path = output_dir / "pc1_pc10_ordered_nested_cv_r2_barplot.png"

plt.savefig(pdf_path, bbox_inches="tight")
plt.savefig(png_path, bbox_inches="tight", dpi=300)

print(f"Saved PC1–PC10 ordered R² bar plot to: {pdf_path}")
print(f"Saved PNG copy to: {png_path}")

plt.show()
plt.close()
```
